```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {"User": [f"User_{i}" for i in range(1, 21)],
    "Followers": [500, 1200, 800, 2500, 3000,
        1500, 4500, 2200, 6000, 3500,
        700, 1800, 5000, 4200, 900,
        2800, 6500, 3200, 1100, 5500],
    "Posts": [50, 120, 80, 200, 250,
        100, 350, 180, 400, 300,
        60, 150, 320, 280, 90,
        220, 450, 260, 110, 380]}

df = pd.DataFrame(data)

sns.scatterplot(data=df,x="Followers",y="Posts")

plt.title("Instagram Followers vs Number of Posts")
plt.xlabel("Number of Followers")
plt.ylabel("Number of Posts")
plt.show()
```

```python
# Task 2.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import random

random.seed(42)

days = list(range(1, 31))
orders = [random.randint(100, 500) for _ in range(30)]

df = pd.DataFrame({"Day": days,
    "Orders": orders})

sns.lineplot(data=df,x="Day",y="Orders",marker="o")

plt.title("Daily Zomato Food Orders")
plt.xlabel("Day of the Month")
plt.ylabel("Number of Food Orders")
plt.xticks(range(1, 31, 2))
plt.show()
```

```python
# Task 3.
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {"movie": ["Movie 1", "Movie 2", "Movie 3", "Movie 4", "Movie 5",
        "Movie 6", "Movie 7", "Movie 8", "Movie 9", "Movie 10",
        "Movie 11", "Movie 12", "Movie 13", "Movie 14", "Movie 15"],
    "rating": [8.5, 7.8, 8.1, 7.2, 9.0,
        7.5, 8.2, 6.9, 8.7, 7.9,
        8.3, 7.1, 9.1, 7.7, 8.0],
    "reviews": [5000, 3200, 4500, 2100, 7000,
        1800, 4000, 1500, 6200, 3500,
        4800, 2200, 7500, 3000, 4200],
    "genre": ["Action", "Comedy", "Drama", "Action", "Drama",
        "Comedy", "Action", "Comedy", "Drama", "Action",
        "Drama", "Comedy", "Action", "Comedy", "Drama"]}

df = pd.DataFrame(data)

sns.relplot(data=df,
    x="rating",
    y="reviews",
    hue="genre",
    kind="scatter",
    s=100)

plt.title("Movie Ratings vs Number of Reviews")
plt.xlabel("Movie Rating")
plt.ylabel("Number of Reviews")
plt.show()
```

```python
# Task 4.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {"YouTube_Hours": [1, 2, 3, 4, 5, 6, 2, 7, 3, 8,
        4, 5, 9, 1, 6, 7, 3, 10, 2, 8,
        5, 4, 9, 6, 7],
    "Exam_Score": [92, 88, 85, 80, 75, 70, 90, 65, 82, 60,
        78, 73, 55, 95, 68, 64, 84, 50, 89, 58,
        72, 79, 53, 67, 62]}

df = pd.DataFrame(data)

sns.regplot(data=df,
    x="YouTube_Hours",
    y="Exam_Score",
    ci=None,
    scatter_kws={"color": "blue"},
    line_kws={"color": "red"})

plt.title("YouTube Usage vs Exam Scores")
plt.xlabel("Hours Spent on YouTube per Day")
plt.ylabel("Exam Score (%)")
plt.show()
```
