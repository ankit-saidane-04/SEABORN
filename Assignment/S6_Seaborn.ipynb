```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Task 1.
df = pd.read_csv("flipkart_product_ratings_prices.csv")

print(df.head())

# Create scatter jointplot
sns.jointplot(x="rating", y="price", data=df, kind="scatter")

plt.show()
```

```python
# Task 2.

df = pd.DataFrame({'delivery_time': [20, 25, 30, 35, 40, 28, 32, 45, 22, 38,
                                     30, 35, 25, 40, 33, 27, 42, 31, 36, 29],
                   'order_value': [150, 200, 250, 300, 350, 220, 280, 400, 180, 320,
                                    260, 310, 190, 360, 270, 210, 380, 240, 330, 230]})

sns.jointplot(x='delivery_time',y='order_value',data=df, kind='hex',cmap='Blues')

plt.show()
```

```python
# Task 3.

df = pd.DataFrame({'likes': [100, 150, 200, 250, 300, 350, 400, 450, 500, 550,
                              180, 220, 320, 380, 420, 480, 520, 600, 280, 360],
                  'comments': [10, 15, 20, 25, 30, 35, 40, 45, 50, 55,
                              18, 22, 32, 38, 42, 48, 52, 60, 28, 36]})

sns.jointplot( x='likes', y='comments', data=df, kind='kde',fill=True,cmap='Blues')

plt.show()
```

```python
# Task 4.

df = pd.DataFrame({'rating': [3.2, 3.5, 3.8, 4.0, 4.2, 4.5, 4.1, 3.9, 4.3, 4.6, 3.7, 4.0, 4.4, 3.6, 4.2, 4.5,
                              3.8, 4.1, 4.3, 4.7],
                'reviews': [50, 100, 150, 200, 300, 500, 250, 180, 400, 600, 120, 220, 450, 90, 320, 550,
                            160, 280, 420, 700]})

sns.jointplot(x='rating', y='reviews',data=df, kind='scatter')
plt.show()

sns.jointplot(x='rating', y='reviews',data=df, kind='hex', cmap='Blues')
plt.show()

sns.jointplot(x='rating', y='reviews',data=df, kind='kde', fill=True, cmap='Blues')
plt.show()
```
