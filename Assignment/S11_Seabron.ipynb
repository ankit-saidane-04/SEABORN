```python
# Task 1.

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

df = pd.DataFrame({"Category": ["Pizza", "Burger", "Biryani", "Momos", "Ice Cream"],
    "Orders": [250, 300, 400, 180, 220]})

sns.set_style("whitegrid")
plt.figure(figsize=(8, 5))

sns.barplot(data=df, x="Category", y="Orders", hue="Category",
            palette="viridis", legend=False)

plt.title("Number of Orders per Food Category")
plt.xlabel("Food Category")
plt.ylabel("Number of Orders")
plt.tight_layout()

plt.savefig("food_orders.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# Task 2.
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

df = pd.DataFrame({"Day": ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"],
    "Steps": [4500, 6200, 7000, 5500, 8500, 11000, 9000]})

sns.set_style("whitegrid")
plt.figure(figsize=(9, 5))

sns.lineplot(data=df, x="Day", y="Steps",
             marker="o", color="blue", linewidth=2)

plt.title("Daily Step Counts for One Week")
plt.xlabel("Day")
plt.ylabel("Number of Steps")
plt.tight_layout()

plt.savefig("weekly_steps.svg", format="svg", bbox_inches="tight")
plt.show()
```

```python
# Task 3.

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

df = pd.DataFrame({"Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],"Sales": [120, 150, 135, 180, 210, 240]})

sns.set_style("darkgrid")
sns.set_palette("Set2")

plt.figure(figsize=(9, 5))

sns.barplot(data=df, x="Month", y="Sales",
            hue="Month", legend=False)

plt.title("Monthly Sales Performance")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.tight_layout()

plt.savefig("monthly_sales_1200dpi.png",
            dpi=1200, bbox_inches="tight")

plt.show()
```

```python
# Task 4.

import seaborn as sns
import matplotlib.pyplot as plt

def apply_presentation_theme():
    sns.set_theme(style="whitegrid",
        context="talk",
        font="DejaVu Sans",
        palette="Set2")

apply_presentation_theme()

apps = ["Instagram", "YouTube", "WhatsApp", "Spotify", "Chrome"]
usage = [30, 25, 20, 15, 10]

plt.figure(figsize=(8, 7))

plt.pie(usage,
    labels=apps,
    autopct="%1.1f%%",
    startangle=90,
    colors=sns.color_palette("Set2", 5))

plt.title("Top 5 Most-Used Apps")
plt.tight_layout()

plt.savefig("top_5_apps.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# Task 5.

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

df = pd.DataFrame({"Order_Value": [200, 350, 500, 250, 450],
    "Delivery_Time": [25, 35, 45, 20, 40],
    "Rating": [4.5, 4.0, 3.5, 4.8, 4.2],
    "Distance": [3, 6, 9, 2, 7]})

corr = df.corr()

sns.set_theme(style="whitegrid")
plt.figure(figsize=(8, 6))

sns.heatmap(corr,annot=True,cmap="coolwarm",fmt=".2f",linewidths=0.5)

plt.title("Food Delivery Correlation Heatmap")
plt.tight_layout()

plt.savefig("food_delivery_heatmap.svg",format="svg",dpi=1200,bbox_inches="tight")

plt.show()
```
