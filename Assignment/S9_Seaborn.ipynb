```python
# Task 1.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({'likes': [100, 200, 300, 400, 500,
              600, 700, 800, 900, 1000],
    'comments': [10, 15, 25, 30, 40,
                 45, 55, 60, 75, 85]})

sns.regplot(data=df, x='likes', y='comments')

plt.title('Instagram Likes vs Comments')
plt.xlabel('Number of Likes')
plt.ylabel('Number of Comments')
plt.show()
```

```python
# Task 2.
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({'price': [100, 150, 200, 250, 300,
              120, 180, 220, 280, 350, 130, 170, 240, 310, 400],
                'delivery_time': [20, 25, 30, 35, 40,22, 28, 32, 38, 45,23, 27, 33, 42, 48]})

sns.lmplot(data=df,x='price',y='delivery_time')

plt.title('Food Order Price vs Delivery Time')
plt.xlabel('Order Price (₹)')
plt.ylabel('Delivery Time (Minutes)')
plt.show()
```

```python
# Task 3.
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({'songs': [5, 10, 15, 20, 25,30, 35, 40, 45, 50,
                             55, 60, 65, 70, 75],
                  'duration': [15, 32, 45, 65, 78,90, 115, 125, 150, 165,
                               180, 205, 220, 240, 260]})

sns.regplot(data=df,x='songs', y='duration',lowess=True,color='green')

plt.title('Spotify Playlist Songs vs Duration')
plt.xlabel('Number of Songs')
plt.ylabel('Total Duration (Minutes)')
plt.show()
```

```python
# Task 4.
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({'hours': [1, 2, 3, 4, 5, 6,
              1, 2, 3, 4, 5, 6],
                  'score': [40, 45, 55, 60, 70, 75,
              45, 55, 65, 75, 80, 90],
                  'method': ['self-study', 'self-study', 'self-study',
        'self-study', 'self-study', 'self-study',
        'coaching', 'coaching', 'coaching',
        'coaching', 'coaching', 'coaching']})

sns.lmplot( data=df, x='hours',y='score',col='method',height=4,aspect=1)

plt.show()
```

```python
# Task 5.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({'price': [500, 1000, 1500, 2000, 2500,
              3000, 3500, 4000, 4500, 5000],
    'reviews': [50, 80, 120, 150, 180,
                220, 250, 280, 320, 350]})

g = sns.lmplot(data=df,
    x='price',
    y='reviews',
    scatter_kws={'color': 'purple'})

g.set_axis_labels('Product Price (₹)','Number of Reviews')

g.figure.suptitle('Flipkart Product Price vs Reviews',y=1.03)

plt.show()
```
