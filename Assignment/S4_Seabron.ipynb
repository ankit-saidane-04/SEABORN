```python
# Task 2.

df = pd.DataFrame({ "Category": ["Fast Food", "Fast Food", "Fast Food", "Fast Food",
        "Fast Food", "Fast Food", "Desserts", "Desserts",
        "Desserts", "Desserts", "Desserts", "Desserts"],
    "Order_Amount": [150, 200, 250, 180, 300, 220, 100, 120, 150, 180, 200, 130]})

# Create violin plot
sns.violinplot(data=df, x="Category", y="Order_Amount",
    inner="quartile")

plt.title("Zomato Order Amount by Restaurant Category")
plt.xlabel("Restaurant Category")
plt.ylabel("Order Amount (₹)")
plt.show()
```

```python
# Task 3.

ticket_prices = [500, 600, 700, 800, 900, 1000, 1100, 1200, 1300, 1500,
    1600, 1800, 2000, 2200, 2500, 3000, 3500, 12000]

sns.boxplot(x=ticket_prices)

plt.title("IPL Match Ticket Price Distribution")
plt.xlabel("Ticket Price (₹)")
plt.show()
```

```python
import numpy as np
Q1 = np.percentile(ticket_prices, 25)
Q3 = np.percentile(ticket_prices, 75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = [price for price in ticket_prices
    if price < lower_limit or price > upper_limit]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
print("Outliers:", outliers)
```
