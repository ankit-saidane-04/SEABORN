```python
# Task 1.

import pandas as pd
import numpy as np

np.random.seed(42)

df = pd.DataFrame({"delivery_time": np.random.randint(15, 60, 20),
    "order_value": np.random.randint(100, 1000, 20),
    "tip_amount": np.random.randint(10, 100, 20),
    "rating": np.random.randint(1, 6, 20),
    "delivery_distance": np.random.uniform(1, 15, 20).round(2)})

print(df.head())

correlation_matrix = df.corr()

print("\nCorrelation Matrix:")
print(correlation_matrix)
```

```python
# Task 2.

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 6))

sns.heatmap(correlation_matrix,annot=True,fmt=".2f", cmap="viridis")

plt.title("Food Delivery Correlation Heatmap")
plt.show()
```

```python
# Task 3.
plt.figure(figsize=(8, 6))

sns.heatmap(correlation_matrix,annot=True,fmt=".2f",cmap="coolwarm",center=0,vmin=-1,vmax=1)

plt.title("Food Delivery Correlation Heatmap - Coolwarm")
plt.show()

```

```python

corr = correlation_matrix.copy()

np.fill_diagonal(corr.values, np.nan)

lower_triangle = corr.where(np.tril(np.ones(corr.shape), k=-1).astype(bool))

pair = lower_triangle.stack().idxmin()
value = lower_triangle.stack().min()

print("Feature 1:", pair[0])
print("Feature 2:", pair[1])
print("Correlation:", round(value, 3))
```

```python
# Task 4.
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

mobile_data = {"price": [12000, 15000, 18000, 22000, 25000,
              28000, 32000, 35000, 40000, 45000],
              "rating": [4.1, 4.2, 4.0, 4.3, 4.4,
               4.2, 4.5, 4.3, 4.6, 4.4],
               "RAM": [4, 4, 6, 6, 8, 8, 8, 12, 12, 16],
               "battery_capacity": [5000, 5000, 5000, 4500, 5000,
                         5000, 5500, 5000, 5000, 5500]}

df_mobile = pd.DataFrame(mobile_data)

print(df_mobile.head())

mobile_corr = df_mobile.corr()

print("\nMobile Phone Correlation Matrix:")
print(mobile_corr)

plt.figure(figsize=(8, 6))

sns.heatmap(mobile_corr, annot=True,fmt=".2f",cmap="RdYlGn",center=0,vmin=-1,vmax=1)

plt.title("Flipkart Mobile Phone Features Correlation")
plt.show()
```
