```python
# Task 1.

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

data = {"Day": ["Monday", "Tuesday", "Wednesday", "Thursday",
            "Friday", "Saturday", "Sunday"],
    "Songs_Streamed": [120, 150, 135, 180, 220, 300, 280]}

df = pd.DataFrame(data)

sns.set(context="poster")

sns.barplot(data=df, x="Day", y="Songs_Streamed")

plt.title("Spotify Daily Songs Streamed")
plt.xlabel("Day")
plt.ylabel("Number of Songs Streamed")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Task 2.

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

data = {"Cuisine": ["Indian", "Chinese", "Italian", "Indian", "Chinese",
        "Indian", "Italian", "Indian", "Chinese", "Italian",
        "Indian", "Chinese", "Indian", "Italian", "Indian",
        "Chinese", "Italian", "Indian", "Chinese", "Indian"]}

df = pd.DataFrame(data)

sns.set_theme(style="whitegrid")

sns.countplot(data=df,
    x="Cuisine",
    palette="colorblind",
    hue="Cuisine",
    legend=False)

plt.title("Zomato Orders by Cuisine Type")
plt.xlabel("Cuisine Type")
plt.ylabel("Number of Orders")
plt.show()
```

```python
# Task 3.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

days = list(range(1, 15))
steps = [4500, 5200, 6100, 4800, 7000, 8500, 7800,
    6200, 9000, 7500, 8200, 9500, 8800, 10000]

df = pd.DataFrame({"Day": days,"Steps": steps})

sns.set(palette="muted",
    context="notebook",
    style="whitegrid")

sns.lineplot(data=df,
    x="Day",
    y="Steps",
    marker="o")

plt.title("Daily Step Counts - 14 Days")
plt.xlabel("Day")
plt.ylabel("Number of Steps")
plt.xticks(days)
plt.tight_layout()
plt.show()
```

```python
# Task 4.
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

df = pd.DataFrame({"Category": ["Electronics", "Clothing", "Groceries", "Books"],"Sales": [500, 350, 420, 200]})

sns.set(palette="deep",context="talk")

sns.barplot(data=df, x="Category", y="Sales", hue="Category", legend=False)
plt.title("Sales by Category - Original")
plt.show()
```

```python
# Updated theme
sns.set(palette="colorblind",context="poster")

sns.barplot(data=df, x="Category", y="Sales", hue="Category", legend=False)
plt.title("Sales by Category - Updated")
plt.show()
```

```python
# Task 5.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {"Category": ["Mobiles", "Laptops", "Fashion","Home Appliances", "Books"],
        "Sales": [150000, 120000, 95000, 110000, 45000]}

df = pd.DataFrame(data)

sns.set(context="talk",
    style="whitegrid",
    palette="deep",
    font_scale=1.1,
    rc={"grid.linestyle": "--",
        "grid.alpha": 0.5,
        "axes.titleweight": "bold"})

plt.figure(figsize=(12, 6))

sns.barplot(data=df,
    x="Category",
    y="Sales",
    hue="Category",
    legend=False)

plt.title("Flipkart Sales Dashboard")
plt.xlabel("Product Category")
plt.ylabel("Total Sales (₹)")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```
