```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Task 1.

delivery_times = [32, 28, 45, 30, 36, 40, 29, 34, 38, 31, 27, 41, 33, 35, 39, 37, 44, 30, 29, 32, 36, 28, 43, 31, 34, 38, 40, 33, 35, 42]

sns.histplot(data=delivery_times, bins=8, kde=False)

plt.title("Distribution of Swiggy Delivery Times")
plt.xlabel("Delivery Time (minutes)")
plt.ylabel("Number of Orders")
plt.show()
```

```python
# Task 2.

steps = [4500, 6200, 7800, 5100, 9000, 8500, 7200, 6800, 10000, 7500, 8200, 5600, 6400, 9500, 11000, 8700, 7300, 6900, 8000, 9200,
         4800, 6000, 10500, 8800, 7600, 9800, 7100, 8400, 9300, 6500]

sns.kdeplot(data=steps, fill=True)

plt.title("Distribution of Daily Step Counts")
plt.xlabel("Number of Steps")
plt.ylabel("Probability Density")
plt.show()
```

```python
# Task 3.

spending = [250, 400, 350, 600, 450, 300, 700, 550, 800, 450, 350, 500, 650, 400, 900, 750, 300, 550, 600, 450,
            1000, 850, 500, 350, 700, 650, 400, 550, 750, 500]

plt.figure(figsize=(8, 5))

sns.histplot(data=spending, bins=8, stat="density", alpha=0.5)
sns.kdeplot(data=spending, color="red", linewidth=2)

plt.title("Distribution of Monthly Zomato Spending")
plt.xlabel("Spending Amount (₹)")
plt.ylabel("Probability Density")
plt.show()
```

```python
# Task 4.

runs = [125, 145, 167, 182, 156, 198, 210, 175, 143, 189,
    201, 160, 155, 220, 178, 134, 190, 165, 205, 148,
    172, 185, 230, 195, 158, 140, 215, 180, 169, 200]

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, bins in zip(axes, [5, 10, 20]):
    sns.histplot(data=runs, bins=bins, ax=ax)

    ax.set_title(f"IPL Scores - {bins} Bins")
    ax.set_xlabel("Runs")
    ax.set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```
